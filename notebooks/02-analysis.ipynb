```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
```

```python
# # Đọc dữ liệu đã làm sạch
df = pd.read_csv('../data/processed/Cleaned_Global_Cybersecurity_Threats_2015-2024.csv')
```

```python
# Số lượng sự cố theo loại tấn công
attack_type_counts = df['Attack Type'].value_counts()
print("\nAttack type counts:")
print(attack_type_counts)
```

```python
# Biểu đồ phân bố loại tấn công
plt.figure(figsize=(12, 6))
attack_type_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Attack Types')
plt.ylabel('')
plt.show()
```
